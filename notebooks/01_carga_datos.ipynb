```python
# 01 — Carga de datos
## Descarga de precios históricos del Panel Líder BYMA (2025)
##Fuente: Yahoo Finance vía yfinance. Datos cargados a PostgreSQL local.
```

```python
import yfinance as yf
import pandas as pd
import numpy as np
from sqlalchemy import create_engine
from dotenv import load_dotenv
import os

load_dotenv(r"C:\proyectos\mercado-capitales\.env")
load_dotenv(r"C:\proyectos\mercado-capitales\.env", override=True)

db_url = os.getenv("DATABASE_URL") or os.getenv("DB_URL")
if not db_url:
    raise ValueError("No se encontró DATABASE_URL/DB_URL en el archivo .env")

engine = create_engine(db_url)
```

```python
tickers = ['YPFD.BA', 'GGAL.BA', 'BMA.BA', 'TXAR.BA', 'ALUA.BA']
datos = yf.download(tickers, start='2025-01-01', end='2026-01-01')
print(datos.shape)
```

```python
precios = datos['Close'].copy()
rendimientos = precios.pct_change() * 100
print("precios:", precios.shape)
print("rendimientos:", rendimientos.shape)
```
