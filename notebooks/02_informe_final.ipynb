## Contexto
Este análisis evalúa el comportamiento de las 5 acciones del Panel Líder de BYMA 
durante 2025, respondiendo tres preguntas clave para la toma de decisiones de inversión:

1. ¿Qué acción tuvo el mayor rendimiento?
2. ¿Cuál presentó mayor riesgo (volatilidad)?
3. ¿Qué tan correlacionadas están entre sí? ¿Ofrecen diversificación real?

## Setup — Imports y conexión

```python
import yfinance as yf
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sqlalchemy import create_engine
from dotenv import load_dotenv
import os

load_dotenv()
engine = create_engine(os.getenv("DATABASE_URL"))
```

## 1. Rendimiento total acumulado 2025

```python
rendimientos = pd.read_sql("SELECT * FROM rendimientos_diarios",engine,index_col="Date",parse_dates=["Date"])
precios = pd.read_sql("SELECT * FROM precios_cierre",engine,index_col="Date",parse_dates=["Date"])
rendimiento_total = (precios.iloc[-1] / precios.iloc[0] - 1) *100
rendimiento_total.plot(
    kind='bar',
    figsize = (10,5)

)
plt.savefig('graficos/rendimiento.png', dpi=150, bbox_inches='tight')
plt.title('Rendimiento total 2025 (%)')
plt.close()
```

![Rendimiento total 2025](graficos/rendimiento.png)

## 2. Volatilidad anualizada

```python
volatilidad = rendimientos.std() * np.sqrt(252)
volatilidad.plot(
    kind='barh'
)
plt.savefig('graficos/volatilidad.png', dpi=150, bbox_inches='tight')
plt.title('Volatilidad Porcentual Anualizada')
plt.close()
```

![Volatilidad anualizada 2025](graficos/volatilidad.png)

## 3. Ratio de Sharpe

```python
sharpe = rendimiento_total / volatilidad

sharpe.sort_values().plot(kind='barh', color='steelblue')
plt.title('Ratio de Sharpe 2025')
plt.xlabel('Sharpe')
plt.axvline(x=0, color='red', linestyle='--', linewidth=0.8)
plt.tight_layout()
plt.savefig('graficos/sharpe.png', dpi=150, bbox_inches='tight')
plt.close()
```

![Ratio de Sharpe 2025](graficos/sharpe.png)

## 4. Correlación entre activos

```python
correlaciones = rendimientos.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(
    correlaciones,
    annot=True,        
    fmt='.2f',         
    cmap='coolwarm',  
    vmin=-1, vmax=1    
)
plt.title('Matriz de correlación — Panel Líder BYMA 2025')
plt.tight_layout()
plt.savefig('graficos/correlacion.png', dpi=150, bbox_inches='tight')
plt.close()
```

![Matriz de correlación 2025](graficos/correlacion.png)
